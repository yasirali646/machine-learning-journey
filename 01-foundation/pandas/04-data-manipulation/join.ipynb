Pandas joins are used to combine data from two DataFrames using a index.

Imagine you have two tables: one contains employee details, and another contains their salaries. You can join them to get all the information in one table.

```python
import pandas as pd

employees = pd.DataFrame({
    "employee_id": [1, 2, 3],
    "name": ["Ali", "Sara", "Ahmed"]
})

salaries = pd.DataFrame({
    "employee_id": [1, 2, 4],
    "salary": [50000, 60000, 70000]
})
```

```python
employees.join(salaries)
```

will raise a ValueError because both DataFrames have overlapping column names and .join() tries to align their indexes by default.

#### Option 1: Use `merge()`

If you want to match on the employee_id column, this is the straightforward approach:

```python
pd.merge(
    employees,
    salaries,
    on="employee_id",
    how="left"
)
```

#### Option 2: Set the index, then use `join()`

```python
result = employees.set_index("employee_id").join(
    salaries.set_index("employee_id"),
    how="left"
)

print(result)
```

Here, `set_index()` moves employee_id from an ordinary column into the index. Now `.join()` can match the two DataFrames by their indexes.

### Can `.join()` use a column from the other DataFrame?

Yes! You can use `on=` to match the calling DataFrame's column against the other DataFrame's index.

```python
employees = pd.DataFrame({
    "employee_id": [1, 2, 3],
    "name": ["Ali", "Sara", "Ahmed"]
})

salaries = pd.DataFrame(
    {"salary": [50000, 60000, 70000]},
    index=[1, 2, 4]
)

result = employees.join(
    salaries,
    on="employee_id",
    how="left"
)

print(result)
```

In this example:

- `employees["employee_id"]` provides the matching values.

- `salaries.index` provides the values to match against.

- `how="left"` keeps every employee.

This is a useful feature of `.join()`.

## `merge()` vs `join()` — Final Comparison

| Feature | `pd.merge()` | `DataFrame.join()` |
|---|---|---|
| Default join | `inner` | `left` |
| Default matching | Common columns or indexes, depending on arguments | Indexes |
| Match columns by name | `on="employee_id"` | Usually set the index first |
| Match a column to the other DataFrame's index | Possible | `on="employee_id"` |
| Best starting point | Column-based joins | Index-based joins |
