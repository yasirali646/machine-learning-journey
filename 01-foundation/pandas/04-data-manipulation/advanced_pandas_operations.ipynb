### 1. Concatenation in Pandas

Concatenation means combining two or more DataFrames or Series together.

For example, a company stores January sales and February sales in separate DataFrames. You can combine them into one dataset.

```python
import pandas as pd

jan_sales = pd.DataFrame({
    "product": ["Laptop", "Phone"],
    "sales": [50000, 30000]
})

feb_sales = pd.DataFrame({
    "product": ["Tablet", "Monitor"],
    "sales": [20000, 15000]
})

result = pd.concat([jan_sales, feb_sales], ignore_index=True)

print(result)
```

Here, `pd.concat()` combines the rows, while `ignore_index=True` creates a fresh index from 0.

#### Row-wise concatenation - axis=0

Adds rows vertically. This is the default.

```python
pd.concat([jan_sales, feb_sales], axis=0)
```

#### Column-wise concatenation - axis=1

Adds columns horizontally, aligning rows by index.

```python
pd.concat([jan_sales, feb_sales], axis=1)
```

#### 2. Apply and Map in Pandas
These methods help you transform data by applying a function or mapping values to new values.

#### A. `Series.map()`

```python
employees = pd.DataFrame({
    "name": ["Ali", "Ahmed", "Zoya"],
    "department_id": [1, 2, 1]
})

departments = {
    1: "Engineering",
    2: "Marketing"
}

employees['department'] = employees['department_id'].map(departments)

employees


```

`Notice that map() looks up each code in the dictionary and returns its corresponding name.`

#### B. `DataFrame.apply()`

Use apply() to apply a function to rows or columns of a DataFrame.

Example: Calculate the average marks of each student.

```python
df = pd.DataFrame({
    "name": ["Ali", "Sara", "Ahmed"],
    "math": [80, 90, 70],
    "science": [85, 95, 75]
})

df["average"] = df[['math', 'science']].apply(
    lambda row: row.mean(),
    axis=1
)

df["total"] = df[['math', 'science']].apply(
    lambda row: row.sum(),
    axis=1
)

df["Result"] = df[['math', 'science']].apply(
    lambda row: row.mean(),
    axis=1
).apply(
    lambda row: "Pass" if row >= 80 else "Fail"
)

df
```

The key part is `axis=1`, which means apply the function to each row.

`axis=0`: Apply the function to each column.

`axis=1`: Apply the function to each row.

#### C. `Series.apply()`

This applies a function to each value in a Series.

```python
products = pd.DataFrame({
    "name": ["Laptop", "Mobile", "Keyboard"],
    "price": [5000, 1012, 540]
})

products["discounted_price"] = products['price'].apply(
    lambda price: price * 0.9
)

products
```

Each price is multiplied by 0.9, giving a 10% discount.

#### D. `DataFrame.map()`

In modern Pandas, DataFrame.map() applies a function element by element to an entire DataFrame.

```python
df = pd.DataFrame({
    "A": [1, 2],
    "B": [3, 4]
})

df.map(lambda x: x * 10)
```

### 3. Date and Time Data in Pandas
Date and time operations let you analyze information based on days, months, years, timestamps, and time differences.

For example, you might want to calculate monthly sales, find orders placed after a particular date, or determine delivery time.

#### A. Convert strings to datetime

Dates are sometimes stored as strings. Use pd.to_datetime() to convert them into proper datetime values.

```python
df = pd.DataFrame({
    "order_date": ["2026-01-15", "2026-02-20", "2026-03-10"],
    "sales": [500, 700, 900]
})

df['order_date'] = pd.to_datetime(df["order_date"], )
print(df.dtypes)
```

Now Pandas recognizes order_date as a datetime column rather than ordinary text.

#### B. Extract year, month, and day

Pandas provides the .dt accessor for datetime columns.

```python
df["year"] = df["order_date"].dt.year
df["month"] = df["order_date"].dt.month
df["day"] = df["order_date"].dt.day
df
```

#### C. Filter by date

Suppose you want all orders placed after February 1, 2026.

```python
df[df["order_date"] > "2026-02-01"]
```

#### D. Calculate time differences
Suppose you want to calculate how many days it took to deliver an order.

```python
df = pd.DataFrame({
    "order_date": ["2026-01-01", "2026-01-10"],
    "delivery_date": ["2026-01-05", "2026-01-18"]
})

df["order_date"] = pd.to_datetime(df["order_date"])
df["delivery_date"] = pd.to_datetime(df["delivery_date"])

df["delivery_days"] = (
    df["delivery_date"] - df["order_date"]
).dt.days
```
