### Grouping and Aggregating
Grouping and aggregation are used to summarize large datasets by dividing records into meaningful groups and calculating statistics for each group. Pandas provides `groupby()` and aggregation functions to perform these operations efficiently.

- It organizes records based on one or more columns.
- Calculates summaries such as total, average, minimum, maximum, and count.
- Grouped results can be further sorted, filtered, transformed, or reshaped.
- These operations are commonly used to analyze sales, customers, products, regions, and other categories.

```python
import pandas as pd

data = {
    "Product": [
        "Laptop", "Phone", "Laptop", "Headphones",
        "Phone", "Keyboard", "Headphones", "Laptop"
    ],
    "Category": [
        "Electronics", "Electronics", "Electronics", "Accessories",
        "Electronics", "Accessories", "Accessories", "Electronics"
    ],
    "Region": [
        "North", "South", "North", "East",
        "South", "East", "North", "South"
    ],
    "Quantity": [2, 5, 3, 8, 4, 10, 6, 1],
    "Unit_Price": [60000, 25000, 60000, 3000, 25000, 2000, 3000, 60000],
    "Rating": [4.5, 4.2, 4.7, 4.0, 4.4, 4.1, 4.3, 4.8]
}

df = pd.DataFrame(data)

print(df)
```

#### Aggregating Data
Reduces a collection of values into a summary value. These methods can be applied directly to a Series or DataFrame when an overall summary is required.

- Use aggregation when individual records are not required in the result.
- Different functions answer different analytical questions.
- Aggregation can be performed before or after grouping data.

```python
df['Quantity'].sum()
```

```python
df['Rating'].mean()
```

```python
df['Unit_Price'].median()
```

```python
df['Quantity'].min()
```

```python
df['Quantity'].max()
```

```python
df["Rating"].count()
```

```python
df['Unit_Price'].std()
```

```python
df['Unit_Price'].var()
```

```python
df['Quantity'].agg(
    ['min', 'max', 'sum', 'mean']
)
```

```python
df['Quantity'].agg(
    minmum="min",
    maximum="max"
)
```

```python
df.agg(
    minimum_rating=("Rating", "min"),
    total_quantity = ("Quantity", "sum")
)
```

### Grouping Data
Aggregation becomes more useful when calculations need to be performed separately for different categories. The groupby() method divides the DataFrame into groups based on one or more columns.

- `groupby()` defines how records should be divided.
- An aggregation function can then be applied to each group.
- Grouping does not change the original DataFrame unless the result is assigned back.

Single Column

```python
df.groupby(['Product'])['Quantity'].sum()
```

Multiple Columns

```python
df.groupby(["Category", "Product"])['Unit_Price'].sum()
```

```python
df.groupby(["Category", "Product"])['Unit_Price'].agg(
    Total = ("sum"),
    Maximum = ("max")
)
```

```python
df.groupby(["Category", "Product"]).agg(
    Total = ('Unit_Price', "sum"),
    Maximum = ('Unit_Price', "max"),
    Average = ('Unit_Price', "mean")
)
```
