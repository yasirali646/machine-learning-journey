## What is Pandas?
- Pandas is a Python library used for working with data sets.

- It has functions for analyzing, cleaning, exploring, and manipulating data.

```python
import pandas as pd
```

#### What is a Series?
- A Pandas Series is like a column in a table.

- It is a one-dimensional array holding data of any type.

```python
nums = [1, 7, 8]

pd.Series(nums)
```

#### Create Labels
With the `index` argument, you can name your own labels.

```python
nums = [10, 20, 30]
s = pd.Series(nums, index=['a', 'b', 'c'])
```

```python
s['b']
```

#### Key/Value Objects as Series
You can also use a key/value object, like a dictionary, when creating a Series.

`Note: The keys of the dictionary become the labels.`

```python
days = {"Mon": 1, "Tue": 2, "Wed": 3, "Thur": 4, "Fri": 5, "Sat": 6, "Sun": 7}
pd.Series(days)
```

#### DataFrames
- Data sets in Pandas are usually multi-dimensional tables, called DataFrame.
- Series is like a column, a DataFrame is the whole table.


```python
data = [
    {"calories" : 420, "duration" : 50},
    {"calories" : 240, "duration" : 30},
    {"calories" : 500, "duration" : 60},
]
pd.DataFrame(data)
```

Preferred way to make dataframes

```python
data = {
    "calories" : [420, 240, 500],
    "duration" : [50, 30, 60]
}
df = pd.DataFrame(data)
```

#### Locate Row
As you can see from the result above, the DataFrame is like a table with rows and columns.

Pandas use the `loc` attribute to return one or more specified row(s)


```python
# Refer to the row index:
df.loc[0]
```

Note: This example returns a Pandas Series.

```python
type(df.loc[0])
```

Return row 0 and 1:

```python
# Use a list of indexes:
df.loc[[0, 1]]
```

Note: When using `[ ]`, the result is a Pandas DataFrame.

```python
type(df.loc[[0, 1]])
```

#### Named Indexes
With the `index` argument, you can name your own indexes.

```python
inventory = {
    "name" : ["Laptop", "Mouse", "Keyboard"],
    "price": [540.12, 12.9, 29.5]
}

df = pd.DataFrame(inventory, index=["item1", "item2", "item3"])
print(df)
```

#### Locate Named Indexes
Use the named index in the `loc` attribute to return the specified row(s).

```python
# Refer to the named index:
print(df.loc["item1"])
```

#### Read CSV Files
- A simple way to store big data sets is to use CSV files (comma separated files).
- CSV files contains plain text and is well know format that can be read by everyone including Pandas.

```python
df_csv = pd.read_csv("data.csv")
print(df_csv)
```

#### Read JSON
- Big data sets are often stored, or extracted as JSON.
- JSON is plain text, but has format of an object, and is well known in the world of programming, including Pandas.

```python
df_json = pd.read_json("data.json")
print(df_json)
```

#### Viewing the Data
- One of most used method for getting a quick overview of the DataFrame, is the `head()` method.
- The `head()` method returns the headers and a specified number or rows, starting from the top.

`Note: if the number of rows is not specified, the head() method will return the top 5 rows.`

```python
df = pd.read_csv("data.csv")

df.head(10)
```

```python
df.head()
```

- There is also `tail()` method for viewing the last rows of the DataFrame.
- The `tail()` method returns the headers and a specified number of rows, starting from the bottom.

```python
print(df.tail())
```

#### Info About the Data
The DataFrames object has a method called `info()`, that gives you more information about the data set.

```python
df.info()
```
