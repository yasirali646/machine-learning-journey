### Duplicates

```python
import pandas as pd

data = {
    "name":["Yasir", "Yaseen", "Shahzain", "Amaar", "Yaseen"],
    "age" : [28, 34, 6, 3, 34],
    "city" : ["New York", "Larkana", "London", "Islamabad", "Larkana"]
}

df = pd.DataFrame(data)
```

#### 1. Duplicated
The `duplicated()` method returns a Boolean values for each row:

```python
df.duplicated()
```

Filter the duplicated rows.

```python
df[df.duplicated()]
```

To get the total number of duplicated rows.

```python
df.duplicated().sum()
```

#### Drop Duplicate
The `drop_duplicate()` method in Pandas is design to remove duplicate rows from a DataFrame based on all columns or specific ones. By default, it scans the entire DataFrame and retains the first occurrence of each row and removes any duplicates that follow.

```python
df_cleaned = df.drop_duplicates()
print(df_cleaned)
```

#### 1. Drop Duplicates based on Specific Columns
We can remove duplicates in specific columns using the subset parameter.

```python
data = {
    "name":["Yasir", "Yaseen", "Shahzain", "Amaar", "Yaseen"],
    "age" : [28, 34, 6, 3, 34],
    "city" : ["New York", "Larkana", "London", "Islamabad", "Larkana"]
}

df = pd.DataFrame(data)
df_cleaned = df.drop_duplicates(subset=["name"])
df_cleaned

```

#### 2. Keeping the Last Occurrence of Duplicates
Be default `drop_duplicates` retains the first occurrence of duplicates. If we want to keep the last occurrence we can use `keep=last`.

```python
df_cleaned = df.drop_duplicates(keep="last")

print("Original DataFrame")
print(df)

print("\nCleaned DataFrame (No Duplicate)")
print(df_cleaned)
```

#### 3. Dropping All Duplicates
If we want to remove all rows that are duplicates, we can `keep=False`.

With `keep=False` both occurrences of Alice are removed leaving only the rows with unique values across all columns.

```python
df_cleaned = df.drop_duplicates(keep=False)

print("Original DataFrame")
print(df)

print("\nCleaned DataFrame (No Duplicate)")
print(df_cleaned)
```

#### 4. Modifying the original DataFrame Directly
If we want to modify the DataFrame in place without creating a new DataFrame set `inplace=True`.

```python
df.drop_duplicates(inplace=True)
print(df)
```
