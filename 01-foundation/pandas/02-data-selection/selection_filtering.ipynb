### Selecting and Filtering Data
This operation retrieves specific columns, rows or records that match a condition. It allows precise extraction of required information.

```python
import pandas as pd
```

```python
df = pd.read_csv("data.csv")
```

```python
df.head()
```

```python
df['age'] >= 30
```

Returns rows where `age` greater or equal to 30

```python
df[ df['age'] >= 30 ]
```

```python
df[ (df['age'] > 22) & (df['sales'] > 200) ]
```

```python
df[ df['name'].str.contains("Ali")]
```
