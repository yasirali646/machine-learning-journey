#### To create an nd array we can pass a list, tuple or any array-like object into the array() methods, and it will be converted into an nd array.

```python
arr = np.array((1, 2, 3, 4))
type(arr)
```

```python
arr = np.array({1, 2, 3, 4, 1})
arr, type(arr)
```

#### 0-D array or Scalars

```python
arr = np.array(50)
arr
```

#### 1-D array or uni-dimensional array or vector

```python
arr = np.array(

    [1, 2, 3, 4, 5, 6]

)
arr
```

#### 2-D Array or Matrix or 2nd order tensors

```python
arr = np.array(
    [
        [1, 2, 3], [4, 5, 6]
    ]
)
arr
```

#### 3-D Array or 3rd order tensor

```python
arr = np.array(
    [
        [
            [1, 2, 3], [4, 5, 6]
        ], 
        [
            [11, 22, 33], [44, 55, 66]
        ]
    ]
)
arr.ndim
```

#### 4-D Array or 4th order tensor

```python
arr = np.array(
    [
        [
            [
                [1, 2, 3], [4, 5, 6]
            ]
        ]
    ]
)
arr.ndim
```

#### ndim is a method to check the dimensions

```python
a = np.array(42)
b = np.array([1, 2, 3, 4, 5])
c = np.array([[1, 2, 3], [4, 5, 6]])
d = np.array([[[1, 2, 3], [4, 5, 6]], [[1, 2, 3], [4, 5, 6]]])

print(a.ndim)
print(b.ndim)
print(c.ndim)
print(d.ndim)
```

Array Size

```python
a = np.array(42)
b = np.array([1, 2, 3, 4, 5])
c = np.array([[1, 2, 3], [4, 5, 6]])
d = np.array([[[1, 2, 3], [4, 5, 6]], [[1, 2, 3], [4, 5, 6]]])

print(a.size)
print(b.size)
print(c.size)
print(d.size)
```

Array Shape

```python
a = np.array(42)
b = np.array([1, 2, 3, 4, 5])
c = np.array([[1, 2, 3], [4, 5, 6]])
d = np.array([[[1, 2, 3], [4, 5, 6]], [[1, 2, 3], [4, 5, 6]]])

print(a.shape)
print(b.shape)
print(c.shape)
print(d.shape)
```

#### Higher Dimensional Array
When the array is created, you can define the number of dimensions by using the ndmin argument.

```python
arr = np.array([1, 2, 3, 4], ndmin=5)

print(arr)
print('number of dimensions :',arr.ndim)
```
