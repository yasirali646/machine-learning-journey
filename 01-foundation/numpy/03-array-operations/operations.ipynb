### Adding, removing, and sorting elements

```python
import numpy as np
```

```python
x = np.array([1, 35, 5, 4])
y = np.array([12, 34, 33, 98])
```

```python
# It didn't add 100 to the end of the array, it created a new array with 100 at the end. The original array x remains unchanged.
x = np.append(x, 100)
x
```

```python
# It didn't delete the first element of the array, it created a new array without the first element. The original array x remains unchanged.
np.delete(x, 0)  # Delete the first element of the array
```

```python
# Update the value at index 4 to 50
x[4] = 50
x
```

```python
np.sort(y)  # Sort the array y in ascending order
y
```

#### Operations with a Scalar
Operation on the entire array.

```python
arr = np.array([2, 3, 5, 9])

print("Addition:", arr + 5)
print("Subtraction:", arr - 2)
print("Multiplication:", arr * 10)
print("Division:", arr / 4)
print("Floor Division:", arr // 3)
print("Modulus:", arr % 2)
print("Power:", arr ** 2)
```

#### Mathematical Functions

```python
arr = np.array(np.arange(1, 11))
```

```python
print("Cummulative Sum:", np.cumsum(arr))
```

```python
print("Cummulative product:", np.cumprod(arr))
```

```python
arr = np.array([10, 15, 25, 5])

"""Differences
A discrete difference means subtracting two successive elements.

E.g. for [1, 2, 3, 4], the discrete difference would be [2-1, 3-2, 4-3] = [1, 1, 1]"""
print("Difference:", np.diff(arr))


"""We can perform this operation repeatedly by giving parameter n.

E.g. for [1, 2, 3, 4], the discrete difference with n = 2 would be [2-1, 3-2, 4-3] = [1, 1, 1] , 
then, since n=2, we will do it once more, with the new result: [1-1, 1-1] = [0, 0]"""
print("Difference:", np.diff(arr, n=2))
# Returns: [5 -30] because: 15-10=5, 25-15=10, and 5-25=-20 AND 10-5=5 and -20-10=-30

```

```python
print("Square Root:", np.sqrt(arr))
```

```python
print("Exponential:", np.exp(arr))
```

```python
print("Log:", np.log(arr))
```

```python
print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Standard Deviation:" ,np.std(arr))
print("Variance", np.var(arr))
```

#### Axis Operation

```python
arr = np.array(np.arange(1, 7)).reshape(2, 3)
arr
```

```python
print("Sum of Everything:", np.sum(arr))

print("Sum of columns:", np.sum(arr, axis = 0))

print("Sum of rows:", np.sum(arr, axis = 1))
```

The same concept works with:

```python
np.mean(arr, axis=0)
np.mean(arr, axis=1)

np.max(arr, axis=0)
np.max(arr, axis=1)

np.min(arr, axis=0)
np.min(arr, axis=1)

np.prod(arr, axis=0)
np.prod(arr, axis=1)
```

#### Where
- Returns the indices where the condition is true.

```python
arr = np.array(np.arange(10, 100, 5))
arr
```

```python
np.where( arr > 50)
```

You can also replace values conditionally:

```python
np.where( arr < 50, arr, arr ** 2)
```

```python
np.where( arr % 2 == 0, "Even", "Odd")
```
